````python
from langchain.agents import initialize_agent, Tool, AgentType
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_experimental.utilities import PythonREPL
from langchain_openai import ChatOpenAI
import textwrap
````

````python
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.1)
````

````python
search = DuckDuckGoSearchRun()
search_tool = Tool(
    name="DuckDuckGo Search",
    func=search.run,
    description="Used to find current economic data, demand forecasts, and weather conditions."
)
````

````python
python_repl = PythonREPL()
python_tool = Tool(
    name="Python Interpreter",
    func=python_repl.run,
    description="Used for calculations, trends, and sales forecasts. Accepts only Python code."
)
````

````python
os.environ["SERPAPI_API_KEY"] = creds["SERPAPI_API_KEY"]
````

````python
tools = load_tools(["serpapi"], llm=llm)
prompt = hub.pull("hwchase17/react")
````

````python
tools.append(search_tool)
tools.append(python_tool)
````

````python
agent = create_react_agent(llm, tools, prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)
````

````python
query = """
Ми експортуємо апельсини з Бразилії.
У 2021 році — 200 тонн, у 2022 — 190 тонн, у 2023 — 210 тонн, у 2024 (який ще не закінчився) — 220 тонн.
Зроби прогноз, скільки ми зможемо експортувати у 2025 році, враховуючи:
- погодні умови в Бразилії,
- рівень інфляції у світі,
- попит на апельсини на глобальному ринку.

Використай пошук в інтернеті для отримання актуальної інформації та Python для розрахунку тренду.
"""
````

````python
agent_executor.invoke({'input': query})
````

````python
# Данні експорту апельсинів за роками
export_data = [200, 190, 210, 220]  # в тоннах за 2021, 2022, 2023, 2024

# Розрахунок середнього зростання
growth_rate = (export_data[-1] - export_data[0]) / (len(export_data) - 1)

# Прогноз на 2025 рік
predicted_export_2025 = export_data[-1] + growth_rate
predicted_export_2025
````
